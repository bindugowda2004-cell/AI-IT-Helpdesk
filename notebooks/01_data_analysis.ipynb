```python
import pandas as pd

df = pd.read_csv("../data/raw/tickets.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
# Check the dataset structure

print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nFirst 5 rows:")
df.head()
```

```python
# Basic statistical summary of the dataset

print("Statistical Summary:")
display(df.describe(include="all"))
```

```python
# Check how many tickets belong to each category

print("Tickets by Category:")
print(df["category"].value_counts())
```

```python
# Check the distribution of ticket priorities

print("Tickets by Priority:")
print(df["priority"].value_counts())
```

```python
# Check the distribution of ticket statuses

print("Tickets by Status:")
print(df["status"].value_counts())
```

```python
import matplotlib.pyplot as plt

# Count tickets in each category
category_counts = df["category"].value_counts()

# Create bar chart
category_counts.plot(kind="bar")

plt.title("Number of Tickets by Category")
plt.xlabel("Category")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Visualize ticket priorities

priority_counts = df["priority"].value_counts()

priority_counts.plot(kind="bar")

plt.title("Number of Tickets by Priority")
plt.xlabel("Priority")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Visualize ticket status

status_counts = df["status"].value_counts()

status_counts.plot(kind="bar")

plt.title("Number of Tickets by Status")
plt.xlabel("Status")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Check for duplicate rows

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Check for missing values in every column

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
# Remove duplicate rows

df = df.drop_duplicates()

print("Dataset shape after removing duplicates:")
print(df.shape)
```

```python
# Check the current data types

print("Current data types:")
print(df.dtypes)
```

```python
# Convert date columns to datetime format

df["created_at"] = pd.to_datetime(df["created_at"])
df["resolved_at"] = pd.to_datetime(df["resolved_at"], errors="coerce")

print("Date columns converted successfully!")

print("\nData types after conversion:")
print(df[["created_at", "resolved_at"]].dtypes)
```

```python
# Create time-based features

df["created_hour"] = df["created_at"].dt.hour
df["created_day"] = df["created_at"].dt.day_name()
df["created_day_number"] = df["created_at"].dt.dayofweek
df["created_month"] = df["created_at"].dt.month

print("Time-based features created successfully!")

print(
    df[
        [
            "created_at",
            "created_hour",
            "created_day",
            "created_day_number",
            "created_month"
        ]
    ].head()
)
```

```python
# Create text-based features

df["description"] = df["description"].fillna("")

df["text_length"] = df["description"].str.len()
df["word_count"] = df["description"].str.split().str.len()

df["contains_urgent"] = (
    df["description"]
    .str.lower()
    .str.contains("urgent")
    .astype(int)
)

df["contains_error"] = (
    df["description"]
    .str.lower()
    .str.contains("error")
    .astype(int)
)

df["contains_password"] = (
    df["description"]
    .str.lower()
    .str.contains("password")
    .astype(int)
)

df["contains_vpn"] = (
    df["description"]
    .str.lower()
    .str.contains("vpn")
    .astype(int)
)

print("Text-based features created successfully!")

print(
    df[
        [
            "description",
            "text_length",
            "word_count",
            "contains_urgent",
            "contains_error",
            "contains_password",
            "contains_vpn"
        ]
    ].head()
)
```

```python
# Analyze resolution time

resolved_tickets = df[df["status"] == "Resolved"]

print("Number of resolved tickets:", len(resolved_tickets))

print("\nResolution time statistics:")
print(resolved_tickets["resolution_time_hours"].describe())
```

```python
# Average resolution time by ticket category

average_resolution = (
    resolved_tickets
    .groupby("category")["resolution_time_hours"]
    .mean()
    .sort_values()
)

print("Average resolution time by category:")
print(average_resolution)

average_resolution.plot(kind="bar")

plt.title("Average Resolution Time by Category")
plt.xlabel("Category")
plt.ylabel("Average Resolution Time (Hours)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Analyze resolution time by priority

priority_resolution = (
    resolved_tickets
    .groupby("priority")["resolution_time_hours"]
    .mean()
    .reindex(["Low", "Medium", "High"])
)

print("Average resolution time by priority:")
print(priority_resolution)

priority_resolution.plot(kind="bar")

plt.title("Average Resolution Time by Priority")
plt.xlabel("Priority")
plt.ylabel("Average Resolution Time (Hours)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Final data-quality check

print("Final dataset shape:")
print(df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```
